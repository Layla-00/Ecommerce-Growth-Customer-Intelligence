```python
import pandas as pd

# قراءة الملفات من مجلد data
users = pd.read_csv('../data/users.csv')
orders = pd.read_csv('../data/orders.csv')
products = pd.read_csv('../data/products.csv')

# عرض حجم البيانات (عدد الصفوف والأعمدة) لكل جدول
print("--- Data Shapes ---")
print(f"Users shape: {users.shape}")
print(f"Orders shape: {orders.shape}")
print(f"Products shape: {products.shape}\n")

# عرض أول 5 صفوف من جدول الطلبات
print("--- Orders Preview ---")
display(orders.head())
```

```python
# 1. التحقق من وجود قيم مفقودة في الجداول
print("--- Missing Values ---")
print("Users Nulls:", users.isnull().sum().sum())
print("Orders Nulls:", orders.isnull().sum().sum())
print("Products Nulls:", products.isnull().sum().sum())

# 2. التحقق من القيم المكررة
print("\n--- Duplicates ---")
print("Users Duplicates:", users.duplicated().sum())
print("Orders Duplicates:", orders.duplicated().sum())
print("Products Duplicates:", products.duplicated().sum())

# 3. تحويل صيغة التاريخ لـ Datetime
orders['order_date'] = pd.to_datetime(orders['order_date'])

print("\n--- Order Date Info ---")
orders.info()
```

```python
from sqlalchemy import create_engine

# 1. إعداد الاتصال بقاعدة بيانات PostgreSQL
# الصيغة: postgresql://username:password@localhost:5432/database_name
engine = create_engine('postgresql://postgres:1415@localhost:5432/postgres')

# 2. رفع الجداول إلى pgAdmin
users.to_sql('users', engine, if_exists='replace', index=False)
orders.to_sql('orders', engine, if_exists='replace', index=False)
products.to_sql('products', engine, if_exists='replace', index=False)

print("تم رفع الجداول بنجاح إلى pgAdmin!")
```
